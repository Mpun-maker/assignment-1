## Types and Memory

```python
# Memory usage before converting to category
before_subject = scores["subject"].memory_usage(deep=True)
before_city = scores["city"].memory_usage(deep=True)

print("Memory usage BEFORE conversion:")
print(f"subject: {before_subject} bytes")
print(f"city: {before_city} bytes")


# Convert subject and city to category dtype
scores["subject"] = scores["subject"].astype("category")
scores["city"] = scores["city"].astype("category")


# Memory usage after converting to category
after_subject = scores["subject"].memory_usage(deep=True)
after_city = scores["city"].memory_usage(deep=True)

print("\nMemory usage AFTER conversion:")
print(f"subject: {after_subject} bytes")
print(f"city: {after_city} bytes")


# Calculate percentage memory saved
subject_saved = ((before_subject - after_subject) / before_subject) * 100
city_saved = ((before_city - after_city) / before_city) * 100

print("\nPercentage memory saved:")
print(f"subject: {subject_saved:.2f}%")
print(f"city: {city_saved:.2f}%")


# Category is cheaper because pandas stores repeated text values only once
# and represents each row using compact integer-like category codes.
# It may not help when a column has many unique values or very few repeated values,
# because the category metadata can add overhead.
```

```python
# build reporting pipeline

import pandas as pd
import numpy as np
import os


# --------------------------------------------------
# 1. LOAD
# --------------------------------------------------

def load(path):
    """Load the raw CSV file."""
    return pd.read_csv(path)


# --------------------------------------------------
# 2. CLEAN
# --------------------------------------------------

def clean(df):
    """Clean the raw scores DataFrame."""

    df = df.copy()

    # Clean inconsistent city names
    df["city"] = (
        df["city"]
        .str.strip()
        .str.title()
    )

    # Fill missing attendance with the column mean
    attendance_mean = df["attendance_percent"].mean()

    df["attendance_percent"] = (
        df["attendance_percent"]
        .fillna(attendance_mean)
    )

    # Replace impossible marks with NaN.
    # Marks outside 0–100 are invalid and should not be
    # used as real exam scores.
    df.loc[
        (df["marks"] < 0) | (df["marks"] > 100),
        "marks"
    ] = np.nan

    # Drop rows where marks are missing
    # because marks are required for the marks analysis.
    df = df.dropna(subset=["marks"])

    # Remove fully duplicated rows
    df = df.drop_duplicates(keep="first")

    # Convert exam_date to datetime
    df["exam_date"] = pd.to_datetime(df["exam_date"])

    # Create exam month
    df["exam_month"] = df["exam_date"].dt.month

    # Convert categorical columns
    df["subject"] = df["subject"].astype("category")
    df["city"] = df["city"].astype("category")

    return df


# --------------------------------------------------
# 3. SUMMARISE
# --------------------------------------------------

def summarise(df):
    """Create the C12(a) summary table."""

    summary = (
        df.groupby(
            ["city", "subject"],
            observed=True
        )["marks"]
        .agg(
            mean_marks="mean",
            max_marks="max",
            count_marks="count"
        )
        .reset_index()
    )

    return summary
```

Why observed=True?

Because city and subject are now categorical columns. observed=True tells pandas to show only the city–subject combinations that actually occur in the cleaned data, rather than potentially generating unused category combinations

```python
# test the entire pipeline 

summary = summarise(clean(load(r"..\data\scores_raw.csv")))

print(summary)
```

Two cleaning decisions could affect the conclusions in my C12 summary. First, I filled missing attendance_percent values with the overall mean. This keeps more students in the dataset, but it assumes that students with missing attendance have attendance similar to the average student. If students with missing attendance were actually much more or much less likely to perform well, mean imputation could weaken or hide the relationship between attendance and marks. Although attendance is not directly aggregated in C12, retaining these rows could affect other conclusions drawn from the cleaned dataset.

Second, I replaced impossible marks outside the 0–100 range with missing values and then dropped rows with missing marks. This prevents invalid scores from increasing or decreasing the city-and-subject means, but it also removes those students completely from the C12 summary. If the invalid values were actually data-entry errors concentrated in a particular city or subject, dropping them could change that group's mean marks and sample count. Therefore, the C12 results describe the cleaned valid-score records rather than necessarily representing every original record. These decisions make the analysis more consistent, but they can still introduce bias if the missing or invalid values are not randomly distributed.
