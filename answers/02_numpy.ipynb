##  Vectorization a python  loop

Convert 
temps
 to Fahrenheit two ways: with an explicit loop (the only loop allowed in Part B) and with one
vectorized expression. Confirm they agree with 
np.allclose 
. Then time both on an array of 1,000,000
values using

, and print both timings plus the speedup as a multiple (

```python
import time

# convert temps to fahrenheit using an explicit loop

temps_flat = temps.ravel()

Fahrenheit_loop = np.empty_like(temps_flat)

start_loop = time.perf_counter()

for i in range(temps_flat.size):
    Fahrenheit_loop[i] = temps_flat[i] * 9/5 + 32

end_loop = time.perf_counter()

loop_time = end_loop - start_loop

Fahrenheit_loop = Fahrenheit_loop.reshape(temps.shape)


print('Fahrenheit using Python loop:')
print(Fahrenheit_loop)
```

```python
# convert temps to Fahrenheit using vectorization

start_vectorized = time.perf_counter()

fahrenheit_vectorized = temps * 9/5 + 32

end_vectorized = time.perf_counter()

vectorized_time = end_vectorized - start_vectorized

print(fahrenheit_vectorized)

```

```python
# confirm both methods agree

print('Loop and vectorized results agree:',
      np.allclose(
          Fahrenheit_loop,
          fahrenheit_vectorized,
          equal_nan=True
      ))
```

```python
# timing on 1,000,000 values  

large_temps = np.random.randint(
    14, 
    36,
    size= 1_000_000
    ).astype(float)

# loop on timing
large_loop_result = np.empty_like(large_temps)

start_loop_large = time.perf_counter()

for i in range(large_temps.size):
    large_loop_result[i] = large_temps[i] * 9/5 + 32


end_loop_large = time.perf_counter()

large_loop_time = end_loop_large - start_loop_large


# vectorized time

start_vectorized_large = time.perf_counter()

large_vectorized_result = large_temps * 9/5 + 32

end_vectorized_large = time.perf_counter()

large_vectorized_time = end_vectorized_large - start_vectorized_large



# calculate speeding

speeding = large_loop_time / large_vectorized_time


print('Python loop timing:', large_vectorized_time, 'seconds')

print('Vectorized timing:', large_vectorized_time, 'seconds')

print('Vectorized speeding:', speeding, 'x')




```

```python

# The Python loop spends extra time repeatedly executing Python-level
# indexing, iteration, and assignment for every individual element,
# whereas vectorization performs the operation inside optimized NumPy
# code over the entire array.
```

```python
# B6 — Broadcasting


# --------------------------------------------------
# (a) City-specific fortnight averages and anomalies
# --------------------------------------------------

city_avg = np.nanmean(
    temps,
    axis=1,
    keepdims=True
)

print("B6a — City fortnight averages:")
print(city_avg)
print("B6a — Average shape:", city_avg.shape)


anomalies = temps - city_avg

print("B6a — Anomaly grid:")
print(anomalies)
print("B6a — Anomaly shape:", anomalies.shape)

print(
    "B6a — Anomaly shape is (5, 14):",
    anomalies.shape == (5, 14)
)


# --------------------------------------------------
# (b) Per-day sensor correction
# --------------------------------------------------

sensor_correction = np.array([
    0.2, -0.1, 0.3, 0.0, -0.2, 0.1, -0.3,
    0.2, 0.0, -0.1, 0.3, -0.2, 0.1, 0.0
])

corrected_temps = temps + sensor_correction

print("B6b — Sensor correction:", sensor_correction)
print("B6b — Corrected temperatures:")
print(corrected_temps)
print("B6b — Corrected temperatures shape:", corrected_temps.shape)


# --------------------------------------------------
# (c) Demonstrate incompatible broadcasting
# --------------------------------------------------

try:
    bad_result = temps + np.array([1, 2, 3, 4, 5])

except ValueError as error:
    print("B6c — Broadcasting ValueError:", error)


# Broadcasting aligns dimensions from the right.
# temps has shape (5, 14), while (5,) aligns with the 14-column
# dimension, so 5 cannot match 14; the correction in (b) succeeds
# because its shape is (14,), which matches the last dimension.
```

##  Daily comfort score

```python
# B7b — Daily comfort score


# Mean temperature across cities for each day
daily_mean_temp = np.nanmean(
    temps,
    axis=0
)


# Mean humidity across cities for each day
daily_mean_humidity = np.nanmean(
    humidity,
    axis=0
)


# Combine the two daily features into a (14, 2) array
features = np.column_stack(
    (daily_mean_temp, daily_mean_humidity)
)


# Define weights
weights = np.array([0.6, -0.1])


# Calculate one comfort score per day
comfort_score = features @ weights


print("B7b — features shape:", features.shape)
print("B7b — weights shape:", weights.shape)
print("B7b — comfort score:", comfort_score)
print("B7b — comfort score shape:", comfort_score.shape)


# The inner dimensions must match because matrix multiplication combines
# each row of features with the weight vector; (14, 2) @ (2,) produces
# one dot product for each of the 14 rows, giving a result of shape (14,).
```

## Reproducible randomness and train/test split

```python
# B9 — Reproducible randomness and train/test split

# Flatten temps from (5, 14) into 70 samples
temps_flat = temps.ravel()

print("B9a — Flattened temps shape:", temps_flat.shape)


# Create an index array containing 0 through 69
indices = np.arange(temps_flat.size)


# Set seed and shuffle
np.random.seed(42)
np.random.shuffle(indices)


# Split 80/20
split_point = int(0.8 * len(indices))

train_indices = indices[:split_point]
test_indices = indices[split_point:]


print("B9b — Training index shape:", train_indices.shape)
print("B9b — Test index shape:", test_indices.shape)

print(
    "B9c — Total number of indices:",
    train_indices.size + test_indices.size
)

print(
    "B9c — Total is 70:",
    train_indices.size + test_indices.size == 70
)


# Confirm that train and test contain no common index
shared_indices = np.intersect1d(
    train_indices,
    test_indices
)

print("B9d — Shared indices:", shared_indices)
print(
    "B9d — Train and test share no index:",
    shared_indices.size == 0
)


# Repeat the exact same process
np.random.seed(42)

indices_repeat = np.arange(temps_flat.size)
np.random.shuffle(indices_repeat)

train_indices_repeat = indices_repeat[:split_point]
test_indices_repeat = indices_repeat[split_point:]


# Confirm that the repeated split is identical
print(
    "B9e — Training split reproducible:",
    np.array_equal(train_indices, train_indices_repeat)
)

print(
    "B9e — Test split reproducible:",
    np.array_equal(test_indices, test_indices_repeat)
)


# Shuffling before splitting prevents the train and test sets from being
# determined by the original ordering of the data, which can introduce
# unwanted patterns or bias. The seed makes the random shuffle repeatable,
# so someone else using the same code and seed can reproduce the same split.
```

```python
assert np.array_equal(train_indices, train_indices_repeat)
assert np.array_equal(test_indices, test_indices_repeat)
```

```python
# B10b — Classifier prediction using argmax

scores = np.array([0.08, 0.62, 0.21, 0.09])
labels = np.array(["rain", "clear", "cloudy", "fog"])

predicted_index = np.argmax(scores)
predicted_label = labels[predicted_index]

print("B10b — Scores:", scores)
print("B10b — Predicted index:", predicted_index)
print("B10b — Predicted label:", predicted_label)

# A classifier needs argmax because argmax returns the position of the
# highest score, which lets us use that position to retrieve the class
# label; max returns only the highest score and loses the class position.
```
